```python
word_counts = {}

for review in X_train:
    words = review.lower().split()

    for word in words:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1

print("Unique words:", len(word_counts))
```

```python
sorted_words = sorted(
    word_counts.items(),
    key=lambda item: item[1],
    reverse=True
)
```

```python
top_words_df = pd.DataFrame(
    sorted_words[:10000],
    columns=["word", "count"]
)

top_words_df["Token ID"] = range(
    1, len(top_words_df) + 1
)

print(top_words_df.head(10))
print("Vocabulary size:", len(top_words_df))
```

```python
word_to_id = {}

for i in range(len(top_words_df)):
    word = top_words_df.iloc[i]["word"]
    token_id = top_words_df.iloc[i]["Token ID"]

    word_to_id[word] = token_id

print("Dictionary size:", len(word_to_id))
print(list(word_to_id.items())[:10])
```

```python
PAD_ID = 0 #fills shorter reviews to a fixed length.
UNK_ID = 10001 ## represents words outside our 10,000-word vocabulary.
```

```python
def review_to_ids(review):
    words=review.lower().split()
    token_ids=[]

    for word in words :
        if word in word_to_id:
            token_ids.append(word_to_id[word])
        else:
            token_ids.append(UNK_ID)
    return token_ids
```

```python
print("Original review:")
print(X_train.iloc[0][:20])
```

```python
## Now testing it on one review 

print(review_to_ids(X_train.iloc[0][:20]))
```

```python
X_train_ids=[]

for review in X_train:
    token_ids=review_to_ids(review)
    X_train_ids.append(token_ids)

print("Total reviews:", len(X_train_ids))
print("First review token IDs:", X_train_ids[0][:20])
print("First review length:", len(X_train_ids[0]))
```

```python
## Restricting every review to 100 words max and padding words with length less than 100 words 

Max_len=100

X_train_padded = []

for token_ids in X_train_ids:
    if len(token_ids) > Max_len:
        token_ids = token_ids[:Max_len]
    else:
        while len(token_ids) < Max_len:
            token_ids.append(PAD_ID)

    X_train_padded.append(token_ids)

print("Total reviews:", len(X_train_padded))
print("First review:", X_train_padded[0])
print("First review length:", len(X_train_padded[0]))
```

```python
y_train = y_train.map({
    "negative": 0,
    "positive": 1
})
```

```python
X_train_array = np.array(X_train_padded, dtype=np.int64)
y_train_array = np.array(y_train, dtype=np.int64)

print("X shape:", X_train_array.shape)
print("y shape:", y_train_array.shape)
print("X data type:", X_train_array.dtype)
print("y data type:", y_train_array.dtype)
```
