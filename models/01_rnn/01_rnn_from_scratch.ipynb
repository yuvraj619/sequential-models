```python
PAD_ID = 0 #fills shorter reviews to a fixed length.
UNK_ID = 10001 ## represents words outside our 10,000-word vocabulary.
```

```python
def review_to_ids(review):
    words=review.lower().split()
    token_ids=[]

    for word in words :
        if word in word_to_id:
            token_ids.append(word_to_id[word])
        else:
            token_ids.append(UNK_ID)
    return token_ids
```

```python
def forward_pass(token_ids):
    h = np.zeros(hidden_size)
    processed_token_ids = []
    
    hidden_states = [h.copy()]
    embeddings = []

    for t in range(len(token_ids)):
        token_id = token_ids[t]

        if token_id == PAD_ID:
            continue
            
        processed_token_ids.append(token_id)
        
        x_t = E[token_id]
        embeddings.append(x_t.copy())

        h = np.tanh(
            W_xh @ x_t
            + W_hh @ h
            + b_h
        )

        hidden_states.append(h.copy())

    logits = W_yh @ h + b_y

    cache = {
    "hidden_states": hidden_states,
    "embeddings": embeddings,
    "token_ids": processed_token_ids
}
    return h, logits, cache
```
