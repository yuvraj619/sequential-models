```python
## Initializing the hidden layer of RNN
```

```python
hidden_size=32

W_xh = np.random.randn(32, 32) * 0.01
W_hh = np.random.randn(32, 32) * 0.01
b_h = np.zeros(32)


print("W_xh:", W_xh.shape)
print("W_hh:", W_hh.shape)
print("b_h:", b_h.shape)
```

```python
## initializing the output layer of RNN 
```

```python
output_size=2

W_yh=np.random.randn(2,32)*0.01
b_y=np.zeros(2,)

print("W_yh:", W_yh.shape)
print("b_y:", b_y.shape)
```

```python
## first hidden state calculation 

h = np.zeros(hidden_size)

x_1 = E[X_train_array[0][0]]

h_1 = np.tanh(
    W_xh @ x_1
    + W_hh @ h
    + b_h
)

print("x_1 shape:", x_1.shape)
print("h_1 shape:", h_1.shape)
```

```python
## Second hidden state calculation

x_2 = E[X_train_array[0][1]]

h_2 = np.tanh(W_xh @ x_2+ W_hh @ h_1+ b_h)

print("x_2 shape:", x_2.shape)
print("h_2 shape:", h_2.shape)
```

```python
def forward_pass(token_ids):
    h = np.zeros(hidden_size)
    processed_token_ids = []
    
    hidden_states = [h.copy()]
    embeddings = []

    for t in range(len(token_ids)):
        token_id = token_ids[t]

        if token_id == PAD_ID:
            continue
            
        processed_token_ids.append(token_id)
        
        x_t = E[token_id]
        embeddings.append(x_t.copy())

        h = np.tanh(
            W_xh @ x_t
            + W_hh @ h
            + b_h
        )

        hidden_states.append(h.copy())

    logits = W_yh @ h + b_y

    cache = {
    "hidden_states": hidden_states,
    "embeddings": embeddings,
    "token_ids": processed_token_ids
}
    return h, logits, cache
```

```python
final_hidden, logits, cache = forward_pass(X_train_array[0])

print("Token IDs:", len(cache["token_ids"]))
print("Embeddings:", len(cache["embeddings"]))
print("Hidden states:", len(cache["hidden_states"]))
```

```python
def softmax(logits):
    exp_values = np.exp(logits - np.max(logits))
    probabilities = exp_values / np.sum(exp_values)
    return probabilities
```

```python
probabilities = softmax(logits)

print("Probabilities:", probabilities)
print("Sum of probabilities:", np.sum(probabilities))
```

```python
def calculate_loss(probabilities, true_label):
    loss = -np.log(probabilities[true_label])
    return loss
```

```python
true_label=1

y_one_hot=np.zeros(output_size)
y_one_hot[true_label]=1
delta_z=probabilities-y_one_hot
print("Delta z :",delta_z)
```

```python
def backward_pass(cache,delta_z):
    hidden_states=cache['hidden_states']
    embeddings=cache['embeddings']
    token_ids=cache['token_ids']

    ## weight ,bias & embedding grad matrix initialization
    dW_yh = np.zeros_like(W_yh)
    db_y = np.zeros_like(b_y)

    dW_xh = np.zeros_like(W_xh)
    dW_hh = np.zeros_like(W_hh)
    db_h = np.zeros_like(b_h)

    dE = np.zeros_like(E)

    # grad for loss wrt output params 
    dW_yh=np.outer(delta_z,hidden_states[-1])
    db_yh=delta_z.copy()
    dh=W_yh.T @ delta_z ## this calculate the loss wrt last hidden state 

    for t in reversed(range(len(embeddings))):
        h_t=hidden_states[t+1]
        h_prev=hidden_states[t]
        x_t=embeddings[t]
        token_id=token_ids[t]

        # moving backward from tanh
        da = dh * (1 - h_t**2)

        # Accumulate hidden-layer parameter gradients(they are adding because in the mathematical derivation we have summation for it at every timestep
        dW_xh += np.outer(da, x_t)
        dW_hh += np.outer(da, h_prev)
        db_h += da

        # Calculate and accumulate embedding grads
        dx = W_xh.T @ da
        dE[token_id] += dx

        # Prop grad to the previous hidden state
        dh = W_hh.T @ da ## just like as we did it outside the loop for the final hidden layer
        
    return dW_yh, db_y, dW_xh, dW_hh, db_h,dE
```

```python
def update_parameters(dW_yh, db_y,dW_xh, dW_hh, db_h,dE,learning_rate):
    # Update output-layer parameters
    W_yh[:] -= learning_rate * dW_yh
    b_y[:]  -= learning_rate * db_y

    # Update hidden-layer parameters
    W_xh[:] -= learning_rate * dW_xh
    W_hh[:] -= learning_rate * dW_hh
    b_h[:]  -= learning_rate * db_h

    # Update embedding matrix
    E[:] -= learning_rate * dE

    # Keep padding embedding fixed at zero
    E[PAD_ID] = 0
```

```python
import time
import numpy as np
# Set how many times we want the model to go through the dataset
epochs = 10
learning_rate = 0.01

# Keep the results from each epoch so we can check them later
train_losses = []
train_accuracies = []
val_losses = []
val_accuracies = []

# Keep track of the total training time
training_start_time = time.time()

# Start Training
for epoch in range(epochs):

    # Start timing this epoch and reset the metrics
    epoch_start_time = time.time()
    total_train_loss = 0.0
    correct_predictions = 0

    # Shuffle the reviews so the model doesn't see them in the same order every time
    indices = np.random.permutation(len(X_train_array))

    # Go through each training review one by one
    for step, i in enumerate(indices, start=1):

        # Get the current review and its actual label
        token_ids = X_train_array[i]
        true_label = int(y_train_array[i])

        # Pass the review through the RNN and get the prediction
        final_hidden, logits, cache = forward_pass(token_ids)

        # Convert the logits into probabilities
        probabilities = softmax(logits)

        # Calculate how wrong the prediction is
        loss = calculate_loss(probabilities, true_label)
        total_train_loss += loss

        # Check whether the prediction is correct
        predicted_label = np.argmax(probabilities)

        if predicted_label == true_label:
            correct_predictions += 1

        # Calculate the gradient of the loss with respect to the logits
        delta_z = probabilities.copy()
        delta_z[true_label] -= 1

        # Run backpropagation and calculate all the gradients
        dW_yh, db_y, dW_xh, dW_hh, db_h, dE = backward_pass(cache, delta_z)

        # Use the gradients to update all the model parameters
        update_parameters(dW_yh, db_y,dW_xh, dW_hh, db_h,dE, learning_rate)

        # Show how far we've got every 1000 reviews
        if step % 1000 == 0:
            elapsed = time.time() - epoch_start_time

            print(
                f"Epoch {epoch + 1}/{epochs} | "
                f"Reviews: {step}/{len(X_train_array)} | "
                f"Current Loss: {loss:.4f} | "
                f"Time: {elapsed / 60:.2f} min",
                flush=True)

    # Calculate the average training loss for this epoch
    avg_train_loss = total_train_loss / len(X_train_array)

    # Calculate training accuracy as a percentage
    train_accuracy = (correct_predictions / len(X_train_array)) * 100

    # Save the training results so we can check them later
    train_losses.append(avg_train_loss)
    train_accuracies.append(train_accuracy)

    # Check how the model performs on validation reviews
    total_val_loss = 0.0
    correct_val_predictions = 0

    # Go through the validation reviews without updating the weights
    for i in range(len(X_val_array)):

        # Get the current validation review and its actual label
        token_ids = X_val_array[i]
        true_label = int(y_val_array[i])

        # Get the model's prediction for this review
        final_hidden, logits, cache = forward_pass(token_ids)

        # Convert the logits into probabilities
        probabilities = softmax(logits)

        # Calculate the validation loss
        loss = calculate_loss(probabilities, true_label)
        total_val_loss += loss

        # Check whether the validation prediction is correct
        predicted_label = np.argmax(probabilities)

        if predicted_label == true_label:
            correct_val_predictions += 1

    # Calculate the average validation loss and accuracy
    avg_val_loss = total_val_loss / len(X_val_array)

    val_accuracy = (correct_val_predictions / len(X_val_array)) * 100

    # Save the validation results so we can check them later
    val_losses.append(avg_val_loss)
    val_accuracies.append(val_accuracy)

    # Calculate how long this epoch took
    epoch_time = time.time() - epoch_start_time

    # Calculate the total time spent training so far
    elapsed_time = time.time() - training_start_time

    # Print all the results after this epoch finishes
    print(
        f"\nEpoch {epoch + 1}/{epochs} completed | "
        f"Train Loss: {avg_train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.2f}% | "
        f"Val Loss: {avg_val_loss:.4f} | "
        f"Val Accuracy: {val_accuracy:.2f}% | "
        f"Epoch Time: {epoch_time / 60:.2f} min | "
        f"Total Time: {elapsed_time / 60:.2f} min\n",
        flush=True
    )
```
